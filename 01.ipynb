## 1. 리스트란?

리스트는 여러 개의 데이터를 순서대로 담을 수 있는 상자라고 생각하면 됩니다.
아래 코드를 실행해서 과일 이름을 담은 리스트를 만들어봅시다.

```python
fruits = ["사과", "바나나", "포도"]
print(fruits)
print("리스트 길이:", len(fruits))
```

## 2. 인덱싱과 슬라이싱

리스트 안의 특정 위치에 있는 값을 꺼내거나, 일부 구간만 잘라올 수 있습니다.

```python
print("첫 번째 과일:", fruits[0])
print("마지막 과일:", fruits[-1])
print("앞의 두 개:", fruits[0:2])
```

## 3. 리스트 요소 추가/삭제

```python
fruits.append("딸기")
print("딸기 추가 후:", fruits)

fruits.insert(1, "수박")
print("수박을 1번 위치에 삽입 후:", fruits)

fruits.remove("바나나")
print("바나나 삭제 후:", fruits)

fruits.pop()
print("맨 마지막 요소 삭제 후:", fruits)
```

## 5. 생각해보기 — 데이터 찾기

리스트 안에서 원하는 값을 찾으려면 어떻게 해야 할까요?
가장 간단한 방법은 처음부터 하나씩 확인해보는 것입니다.
이런 방식을 앞으로 '선형 탐색'이라고 부릅니다. 지금은 코드가 어떻게 동작하는지만 눈으로 따라가 봅시다.

```python
numbers = [7, 3, 9, 1, 5, 2, 8]
target = 5

found = False
for num in numbers:
    if num == target:
        found = True
        break

if found:
    print(f"{target}을(를) 찾았습니다!")
else:
    print(f"{target}이(가) 리스트에 없습니다.")
```
